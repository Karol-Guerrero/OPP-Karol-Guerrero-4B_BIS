```python
class proyector:
  def _init_(self, color, size, brand):
    self.color = color
    self.size = size
    self.brand = brand

  def turn-on(self);
    print("The proyector is turn on")

  def describe(self):
    print (f"This table is made with {self.color} & size: {self.size} & and brand: {self.brand}")

#Self: Se utiliza para decirle a python a que objeto pertenecen los atributos
#Create an instance using a class "table"
proyector1 = proyector ("white", "small", "epson")
proyector2 = proyector ("yellow", "big", "epson")

#We access to the object (instance) "proyector1" to call its data
print(proyector1.brand)
print(proyector1.color)
proyector1.turn-on()
proyector1.describe()

#Instance "proyector2"
print(proyector2.brand)
print(proyector2.color)
proyector2.turn-on()
proyector2.describe()

```
